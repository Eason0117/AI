```python
from tensorflow import keras  # 從 TensorFlow 套件中匯入 Keras 深度學習 API
from keras.models import Sequential  # 匯入 Sequential 模型，用來建立循序式神經網路
from keras.layers import Dense, Flatten  # 匯入 Dense 與 Flatten 層
```

```python
mnist=keras.datasets.mnist
(train_data, train_label), (test_data, test_label) = mnist.load_data()
```

```python
print(type(train_data))
print(train_data.shape)
print(type(train_label))
print(train_label.shape)
print(test_data.shape)
print(test_label.shape)
```

```python
print(train_label[0])
print(train_data[0])
```

```python
import matplotlib.pyplot as plt  # 匯入 Matplotlib 的 pyplot 模組，並取別名為 plt 用來繪製圖表及顯示影像
plt.imshow(train_data[3], cmap='binary')
plt.show()  # 使用 show() 將圖片顯示出來
```

```python
model=Sequential()
model.add(Flatten())
model.add(Dense(32, activation='relu', input_dim=784))
model.add(Dense(10, activation='sigmoid'))
model.compile(optimizer='rmsprop', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(train_data,train_label,epochs=20,batch_size=512)

score=model.evaluate(test_data, test_label) # 印出模型的準確率（Accuracy） # score[0]：測試損失值（Loss） # score[1]：測試準確率（Accuracy）
print(score[1])
```
